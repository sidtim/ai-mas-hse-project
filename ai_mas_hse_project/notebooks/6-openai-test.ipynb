```python
from langchain_openai import ChatOpenAI
import os

from dotenv import load_dotenv
load_dotenv()

OPEN_API_KEY_VSE_LLM = os.getenv("OPEN_API_KEY_VSE_LLM")
```

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(
    model="deepseek/deepseek-v3.2", # https://vsellm.ru/models/deepseek/deepseek-v3.2
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2,
    api_key=OPEN_API_KEY_VSE_LLM,
    base_url="https://api.vsellm.ru/v1",
    # organization="...",
    # other params...
)
```

```python
messages = [
    (
        "system",
        "You are a helpful translator. Translate the user sentence to French.",
    ),
    ("human", "I love programming."),
]
```

```python
res = model.invoke(messages)
```
