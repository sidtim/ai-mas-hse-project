```python
import pandas as pd
import numpy as np
import pickle
import seaborn as sns
import matplotlib.pyplot as plt

pd.set_option("display.max_rows", 500)
pd.set_option("display.max_columns", 500)
```

```python
with open("../data/final_dataset/list_dict_with_tasks.pkl", "rb") as file:
    loaded_list = pickle.load(file)
```

```python
df = pd.DataFrame(loaded_list)
df = df.sort_values("id").reset_index(drop=True)

df.head()
```

```python
df['complexity_level_text'] = df['complexity_level'].map({'2-': 'Легкий',
                                                          '2': 'Легкий',
                                                          '2+': 'Средний',
                                                          '3-': 'Средний',
                                                          '3': 'Тяжелый',
                                                          '3+': 'Тяжелый'})
```

```python
(
    df.complexity_level.value_counts().sort_index(),
    df.complexity_level_text.value_counts().sort_index()
)
```

```python
reverted_list = df.to_dict('records')
```

```python
with open("../data/final_dataset/list_dict_with_tasks_update.pkl", "wb") as file:
    pickle.dump(reverted_list, file)
```
